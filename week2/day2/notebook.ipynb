***3- make a numpy arrays for female and male and use Conditional Probability to see the result***

```python
stud=np.array(['male']*60+['female']*40)
grades=np.array(
    ['yes']*30+
    ['no']*30+
    ['yes']*20+
    ['no']*20
)
total=len(stud)
males=stud=='male'
malescount = np.sum(males)
onlymales=grades[males]
pass_male=np.sum(onlymales=='yes')
pOFb=malescount/total
anb=pass_male/total
pOFanb=anb/pOFb
print(pOFanb)
```

## Conditional Probability

We calculated the probability of passing given that the student is male.

The manual calculation was:

P(Pass | Male) = 30 / 60 = 0.5

The Python simulation produced the same result, which verifies the conditional probability calculation.
