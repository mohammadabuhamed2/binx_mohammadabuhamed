# step 1

setup the data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot
from sklearn.model_selection import train_test_split

df=pd.read_csv(r"C:\Users\Hp\Desktop\binx_mohammadabuhamed\binx_mohammadabuhamed\week3\day3\WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.isna().sum()
df.columns
df.dtypes
df['Dependents']= df['Dependents'].map({
    'Yes':1,
    'No':0
})
df['Partner']= df['Partner'].map({
    'Yes':1,
    'No':0
})
df=df[['SeniorCitizen','Partner','Dependents','tenure','MonthlyCharges','Contract','Churn']]
y=df['Churn']
x=df.drop('Churn',axis=1)
x=pd.get_dummies(x)
x.columns
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

### Decision Trees

```python
from sklearn.tree import DecisionTreeClassifier
model= DecisionTreeClassifier(max_depth=5,random_state=42)
model.fit(X_train,y_train)
pred=model.predict(X_test)
```

### Random Forests

```python
from sklearn.ensemble import RandomForestClassifier
model1=RandomForestClassifier(n_estimators=100,random_state=42)
model1.fit(X_train,y_train)
pred1=model1.predict(X_test)
```

### Support Vector Machines (SVM)


```python
from sklearn.svm import SVC
model2=SVC(kernel='rbf',probability=True)
model2.fit(X_train,y_train)
pred2=model2.predict(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model3=KNeighborsClassifier(n_neighbors=5)
model3.fit(X_train,y_train)
pred3=model3.predict(X_test)
```

# step 2

compare by f1

```python
from sklearn.metrics import f1_score

results = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "Random Forest",
        "SVM",
        "k-NN"
    ],
    "F1-score": [
        f1_score(y_test, pred, pos_label="Yes"),
        f1_score(y_test, pred1, pos_label="Yes"),
        f1_score(y_test, pred2, pos_label="Yes"),
        f1_score(y_test, pred3, pos_label="Yes")
    ]
})

results
```

### Model Comparison and Winner Analysis:

"After evaluating all four models on the same test split using the F1-score metric, the Decision Tree model achieved the highest performance with an F1-score of 0.580, outperforming the Random Forest (0.531), SVM (0.523), and k-NN (0.522).

Why it likely won:
Although Ensemble models like Random Forests are generally robust, this specific dataset appears to be governed by clear, direct, and straightforward logical rules that a single Decision Tree can capture efficiently without severe overfitting. On the other hand, distance-based models (SVM and k-NN) struggled slightly more with the dataset's specific feature space, proving the 'no free lunch' principle: the simplest model can sometimes outperform more complex ones depending on the underlying data distribution."
