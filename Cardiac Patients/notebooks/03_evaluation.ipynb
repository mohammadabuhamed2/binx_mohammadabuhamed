# Model Evaluation

## Load the Test Dataset

```python
import pandas as pd

test_data = pd.read_csv(
    r"C:\Users\Hp\Desktop\binx_mohammadabuhamed\binx_mohammadabuhamed\Cardiac Patients\data\test_data.csv"
)
```

```python
X_test = test_data.drop("has_heart_disease", axis=1)
y_test = test_data["has_heart_disease"]
```

## Load the Trained Models

```python
import joblib

logistic_model = joblib.load(
    r"C:\Users\Hp\Desktop\binx_mohammadabuhamed\binx_mohammadabuhamed\Cardiac Patients\models\logistic_regression_pipeline.pkl"
)

svm_model = joblib.load(
    r"C:\Users\Hp\Desktop\binx_mohammadabuhamed\binx_mohammadabuhamed\Cardiac Patients\models\svm_pipeline.pkl"
)

decision_tree_model = joblib.load(
    r"C:\Users\Hp\Desktop\binx_mohammadabuhamed\binx_mohammadabuhamed\Cardiac Patients\models\decision_tree_pipeline.pkl"
)

random_forest_model = joblib.load(
    r"C:\Users\Hp\Desktop\binx_mohammadabuhamed\binx_mohammadabuhamed\Cardiac Patients\models\random_forest_pipeline.pkl"
)

knn_model = joblib.load(
    r"C:\Users\Hp\Desktop\binx_mohammadabuhamed\binx_mohammadabuhamed\Cardiac Patients\models\knn_pipeline.pkl"
)
```

## Make Predictions

```python
logistic_pred = logistic_model.predict(X_test)

svm_pred = svm_model.predict(X_test)

decision_tree_pred = decision_tree_model.predict(X_test)

random_forest_pred = random_forest_model.predict(X_test)

knn_pred = knn_model.predict(X_test)
```

## Logistic Regression Evaluation

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, logistic_pred)

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Heart Disease", "Heart Disease"]
).plot()

plt.title("Logistic Regression Confusion Matrix")
plt.show()
```

### Classification Report

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    logistic_pred,
    target_names=["No Heart Disease", "Heart Disease"]
))
```

### Logistic Regression Evaluation Results

The Logistic Regression model achieved an overall accuracy of **91%** on the unseen test set.

For patients with heart disease, the model achieved:

- **Precision:** 0.87
- **Recall:** 0.83
- **F1-score:** 0.85

The confusion matrix shows:

- **1178 True Negatives:** Patients without heart disease correctly classified as negative.
- **459 True Positives:** Patients with heart disease correctly identified.
- **66 False Positives:** Patients without heart disease incorrectly classified as having heart disease.
- **97 False Negatives:** Patients with heart disease incorrectly classified as not having heart disease.

False Negatives are particularly important in this context because they represent patients who actually have heart disease but were not identified by the model.

Overall, Logistic Regression shows strong classification performance, but its results should be compared with the other trained models before selecting the final model.

## ROC-AUC Evaluation

```python
logistic_prob = logistic_model.predict_proba(X_test)[:, 1]

svm_prob = svm_model.predict_proba(X_test)[:, 1]

decision_tree_prob = decision_tree_model.predict_proba(X_test)[:, 1]

random_forest_prob = random_forest_model.predict_proba(X_test)[:, 1]

knn_prob = knn_model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import roc_auc_score

logistic_auc = roc_auc_score(y_test, logistic_prob)
svm_auc = roc_auc_score(y_test, svm_prob)
decision_tree_auc = roc_auc_score(y_test, decision_tree_prob)
random_forest_auc = roc_auc_score(y_test, random_forest_prob)
knn_auc = roc_auc_score(y_test, knn_prob)

print("Logistic Regression ROC-AUC:", logistic_auc)
print("SVM ROC-AUC:", svm_auc)
print("Decision Tree ROC-AUC:", decision_tree_auc)
print("Random Forest ROC-AUC:", random_forest_auc)
print("KNN ROC-AUC:", knn_auc)
```

### ROC-AUC Results

The ROC-AUC scores for the evaluated models are:

- **Logistic Regression:** 0.9604
- **SVM:** 0.9599
- **Random Forest:** 0.9464
- **KNN:** 0.9214
- **Decision Tree:** 0.9110

Logistic Regression achieved the highest ROC-AUC score, followed very closely by SVM.

Both models demonstrated excellent ability to distinguish between patients with and without heart disease.

Random Forest also achieved strong discrimination performance, while KNN and Decision Tree achieved lower ROC-AUC scores compared with the other models.

Overall, Logistic Regression currently provides the strongest balance between ROC-AUC, F1-score, accuracy, and recall.

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

# Calculate ROC curve for each model
logistic_fpr, logistic_tpr, _ = roc_curve(y_test, logistic_prob)
svm_fpr, svm_tpr, _ = roc_curve(y_test, svm_prob)
tree_fpr, tree_tpr, _ = roc_curve(y_test, decision_tree_prob)
rf_fpr, rf_tpr, _ = roc_curve(y_test, random_forest_prob)
knn_fpr, knn_tpr, _ = roc_curve(y_test, knn_prob)

# Plot ROC curves
plt.figure(figsize=(9, 6))

plt.plot(logistic_fpr, logistic_tpr, label=f"Logistic Regression (AUC = {logistic_auc:.3f})")
plt.plot(svm_fpr, svm_tpr, label=f"SVM (AUC = {svm_auc:.3f})")
plt.plot(tree_fpr, tree_tpr, label=f"Decision Tree (AUC = {decision_tree_auc:.3f})")
plt.plot(rf_fpr, rf_tpr, label=f"Random Forest (AUC = {random_forest_auc:.3f})")
plt.plot(knn_fpr, knn_tpr, label=f"KNN (AUC = {knn_auc:.3f})")

plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()
plt.show()
```

### ROC Curve Analysis

The ROC curves show that all evaluated models perform considerably better than random classification.

Logistic Regression achieved the highest ROC-AUC score of approximately **0.960**, followed very closely by SVM with approximately **0.960**.

Random Forest also demonstrated strong discrimination performance with an ROC-AUC score of **0.946**.

KNN and Decision Tree achieved lower ROC-AUC scores of **0.921** and **0.911**, respectively.

Overall, Logistic Regression and SVM demonstrated the strongest ability to distinguish between patients with and without heart disease across different classification thresholds.

# Final Model Comparison

The evaluated classification models are compared using Accuracy, Precision, Recall, F1-score, ROC-AUC, and the number of False Negatives.

Since failing to identify a patient with heart disease is more critical in this project, particular attention is given to Recall and False Negatives when selecting the final model.

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "SVM",
        "Decision Tree",
        "Random Forest",
        "KNN"
    ],

    "Accuracy": [0.91, 0.91, 0.86, 0.88, 0.85],

    "Precision": [0.87, 0.88, 0.84, 0.79, 0.87],

    "Recall": [0.83, 0.81, 0.68, 0.85, 0.62],

    "F1-score": [0.85, 0.85, 0.75, 0.82, 0.72],

    "ROC-AUC": [
        0.9604,
        0.9599,
        0.9110,
        0.9464,
        0.9214
    ],

    "False Negatives": [97, 103, 180, 84, 214]
})

results
```

## Final Model Selection

Based on the evaluation results, **Random Forest** was selected as the final model.

Although Logistic Regression and SVM achieved higher overall Accuracy and ROC-AUC scores, Random Forest achieved the **highest Recall for the Heart Disease class (0.85)** and produced the **lowest number of False Negatives (84)**.

In this project, False Negatives are considered particularly important because they represent patients who actually have heart disease but are incorrectly predicted as not having the disease.

A False Positive may result in a patient without heart disease being flagged for further evaluation. In contrast, a False Negative means that a patient with heart disease may not be identified by the model.

Therefore, the model selection prioritizes **Recall and minimizing False Negatives** over maximizing Accuracy alone.

### Selected Model

**Random Forest**

- Accuracy: **0.88**
- Precision: **0.79**
- Recall: **0.85**
- F1-score: **0.82**
- ROC-AUC: **0.9464**
- False Negatives: **84**

Random Forest provides a strong balance between overall classification performance and detecting patients with heart disease, while producing the fewest False Negatives among the evaluated models.
