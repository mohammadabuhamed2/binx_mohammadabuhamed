## Step 1 — Plot Activation Functions

### 1-ReLU Activation Function

$$
\text{ReLU}(x)=\max(0,x)
$$
ReLU returns 0 for negative inputs and keeps positive inputs unchanged.

```python
import numpy as np
x=np.linspace(-5,5,100)
relu=np.maximum(0,x)
```

#### Plot relu

```python
import matplotlib.pyplot as plt
plt.plot(x,relu)
plt.plot(x, relu)
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("ReLU Activation Function")
plt.show()
```

#### ReLU Plot Interpretation

The ReLU function outputs 0 for all negative input values.

For positive input values, the output increases linearly and is equal to the input value.

This is why the graph is flat at 0 for negative values and then rises as a straight line for positive values.

ReLU is commonly used in hidden layers because it introduces non-linearity while remaining simple and efficient.

### 2-Sigmoid Activation Function

$$
\text{Sigmoid}(x)=\frac{1}{1+e^{-x}}
$$

Sigmoid transforms any input into a value between 0 and 1.

```python
sigmoid = 1 / (1 + np.exp(-x))

plt.plot(x, sigmoid)
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Sigmoid Activation Function")
plt.show()
```

#### Sigmoid Plot Interpretation

The Sigmoid function transforms any input value into an output between 0 and 1.

Large negative inputs produce values close to 0, while large positive inputs produce values close to 1.

When the input is 0, the output is 0.5.

This S-shaped curve makes Sigmoid suitable for binary classification output, where the result can be interpreted as a probability.

### 3-Tanh Activation Function

$$
\tanh(x)=\frac{e^x-e^{-x}}{e^x+e^{-x}}
$$
Tanh transforms inputs into values between -1 and 1.

```python
tanh = np.tanh(x)

plt.plot(x, tanh)
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Tanh Activation Function")
plt.show()
```

### Tanh Plot Interpretation

The Tanh function transforms input values into a range between -1 and 1.

Large negative inputs approach -1, while large positive inputs approach 1.

When the input is 0, the output is also 0.

The graph has an S-like shape similar to Sigmoid, but it is centered around zero.

## Step 2 — Choose the Output Activation and Loss Function

### 1-Output Activation Function: Sigmoid

The project is a binary classification task because the target variable has two classes: 0 and 1.

Therefore, the **Sigmoid activation** function is used in the output layer because it produces a value between 0 and 1, which can be interpreted as the probability of the positive class.

### 2-Loss Function: Binary Cross-Entropy

**Binary Cross-Entropy** is selected because the project is a binary classification problem.

It compares the predicted probability with the true label and produces a larger loss when the prediction is far from the correct answer.

## Step 3 — Compute a Forward Pass

### Define a Small Neural Network

A small neural network is created with:

- 2 input features
- 2 neurons in one hidden layer
- ReLU activation in the hidden layer
- 1 output neuron
- Sigmoid activation in the output layer

```python
import numpy as np

X_sample = np.array([2.0, 1.0])

W1 = np.array([
    [0.5, -0.2],
    [0.3, 0.8]
])

b1 = np.array([0.1, -0.1])
```

### Calculate the Hidden Layer Weighted Sum

```python
z1 = np.dot(X_sample, W1) + b1

print(z1)
```

Neuron 1:

$$ (2.0 \times 0.5) + (1.0 \times 0.3) + 0.1 $$ $$ = 1.0 + 0.3 + 0.1 $$ $$ = 1.4 $$

Neuron 2:

$$ (2.0 \times -0.2) + (1.0 \times 0.8) - 0.1 $$ $$ = -0.4 + 0.8 - 0.1 $$ $$=0.3$$

### Apply ReLU Activation to the Hidden Layer

```python
a1 = np.maximum(0, z1)

print(a1)
```

### Calculate the Output Layer Weighted Sum

#### Define the Output Layer Weights and Bias

```python
W2 = np.array([0.7, -0.5])
b2 = 0.2
```

#### Calculate the Output Layer Weighted Sum

```python
z2 = np.dot(a1, W2) + b2

print(z2)
```

### Apply Sigmoid Activation to the Output Layer

$$
\text{Sigmoid}(x)=\frac{1}{1+e^{-x}}
$$

```python
output = 1 / (1 + np.exp(-z2))

print(output)
```

### Forward Pass Result

The sample input was passed through the network using the weighted sums, biases, ReLU activation in the hidden layer, and Sigmoid activation in the output layer.

The final output was approximately 0.74.

This value represents the predicted probability of the positive class for this simple example.
