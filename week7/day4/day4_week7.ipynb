### Import the Transformers Pipeline

### Load a Pre-trained Sentiment Analysis Model

```python
classifier = pipeline("sentiment-analysis")
```

### Test Multiple Sentences

```python
texts = [
    "This course is amazing!",
    "I really hate this movie.",
    "The project was difficult but very useful."
]
results = classifier(texts)
print(results)
```

### Sentiment Analysis Results

The pre-trained Transformer successfully classified the sample sentences.

- The first sentence was classified as **Positive** with very high confidence.
- The second sentence was classified as **Negative** with very high confidence.
- The third sentence was classified as **Positive** with high confidence.

This shows that the pre-trained Transformer can understand the overall sentiment of text without training a model from scratch.

## Step 3: Attention vs RNN Memory

### How Attention Differs from RNN Memory

RNNs process sequences step by step and carry information through a hidden state from one time step to the next.

Attention works differently by allowing each element in the sequence to directly examine all other elements and assign higher importance to the most relevant ones.

This direct access helps Transformers capture long-range relationships more effectively and allows sequence positions to be processed in parallel instead of one at a time.

## Step 4: Select the Core Architecture

### Core Model Selection

For the ECG heartbeat classification project, the LSTM will be used as the core model.

The LSTM achieved strong performance on the ECG time-series data, reaching 97.54% test accuracy with a low test loss of 0.0905.

The ECG dataset is sequential, and the order of the signal values is important for recognizing heartbeat patterns.

Although Transformers are powerful for modeling long-range dependencies, the LSTM already performed very well on this dataset and is simpler to train and evaluate for the current project.

Therefore, the LSTM is selected as the main architecture for the ECG heartbeat classification project.
